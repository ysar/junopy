```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
filename = 'fffr_20210608.csv'

df = pd.read_csv(filename)
df = df.drop(df.columns[0], axis=1)
```

```python
fig, axs = plt.subplots(3, 2, figsize=(6, 5), dpi=300)

a = axs[0, 0]
a.hist(df['redchi1'], bins=np.linspace(0, 0.5, 20), color='r', histtype='step')
a.set_xlabel('Step 1 - $\chi^2_r$', weight='bold')
a.text(0.9, 0.8, "a)", fontweight="bold", transform=a.transAxes)
a.set_xlim([0, 0.5])

a = axs[0, 1]
a.hist(df['redchi2'], bins=np.linspace(0, 0.5, 20), color='r', histtype='step')
a.set_xlabel('Step 2 - $\chi^2_r$', weight='bold')
a.text(0.9, 0.8, "b)", fontweight="bold", transform=a.transAxes)
a.set_xlim([0, 0.5])

a = axs[1, 0]
a.hist(df['alpha'].abs(), color='r', histtype='step')
a.set_xlabel('$\\alpha$', weight='bold')
a.text(0.9, 0.8, "c)", fontweight="bold", transform=a.transAxes)

a = axs[1, 1]
a.hist(df['IF'], color='r', histtype='step')
a.set_xlabel('Impact Parameter (IP)', weight='bold')
a.text(0.9, 0.8, "d)", fontweight="bold", transform=a.transAxes)

a = axs[2, 0]
a.hist(df['B0'], color='r', histtype='step')
a.set_xlabel('$B_0$ (nT)', weight='bold')
a.text(0.9, 0.8, "e)", fontweight="bold", transform=a.transAxes)

a = axs[2, 1]
a.hist(df['dtheta'], color='r', histtype='step')
a.set_xlabel('$\Delta \\theta$ = $\cos^{-1}($x$_A$ $\cdot$ x$_M)$ ($^\circ$)', weight='bold')
a.text(0.9, 0.8, "f)", fontweight="bold", transform=a.transAxes)

fig.tight_layout()
fig.savefig('out_images/force-free-flux-rope-histograms.png', bbox_inches='tight')
```
