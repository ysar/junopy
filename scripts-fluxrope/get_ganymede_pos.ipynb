## Get Ganymede location in Sys III

```python
import numpy as np 
import datetime as dt

from JunoPy.spice import InitSpiceKernels, get_ganymede_pos
```

```python
time = dt.datetime(2021, 6, 7, 17, 0)
print(time)
```

```python
InitSpiceKernels()
```

```python
pos = get_ganymede_pos(time)
pos = np.asarray(pos)
print(pos)
```

```python
r = np.sqrt(np.sum(pos**2))
print(r)
```

```python
lon = np.arctan(pos[1]/ pos[0]) * 180 / np.pi
print(lon)
```

```python
lat = np.arccos(pos[2] / r) * 180 / np.pi
print(lat)
```
