# Download data from PDS
Given a `StartDate` and an `EndDate`, download the specified datasets from PDS and save them in the `data` directory.

```python
import datetime as dt
import glob

import JunoPy

from JunoPy.index import IndexFile
from JunoPy.mag import MagData
```

Specify start and end times. 

```python
st = dt.datetime(2016, 7, 8, 17, 30)
et = dt.datetime(2019, 5, 31, 16)
```

Download magnetometer files. All we need to do is instantiate an `IndexFile` object and call the `get_files_between_dates` method. 

```python
magindex = IndexFile(dataset='mag')
magfilelist = magindex.get_files_between_dates(
    st, et, DoDownload=True, sid='SUNSTATE')
```

Convert all `.STS` files into `.hdf5` files. 

```python
for magfile in glob.glob('data/mag/*.sts'):
    print('Converting file - {:s}'.format(magfile), end='\r')
    magdata = MagData(filename=magfile, ishdf=False)
    magdata.convert_to_spherical()
    magdata.save_to_hdf5()
```

Repeat the process to download data from the Waves instrument.

```python
wavesindex = IndexFile(dataset='waves-survey')
wavesfilelist = wavesindex.get_files_between_dates(
    st, et, DoDownload=True, sid='SURVEY')
```
