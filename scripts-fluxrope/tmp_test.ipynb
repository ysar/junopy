```python
import re
import numpy as np

### Read in the streamtraces file and plot the streamlines.

def read_streamtraces_file():
    
    nZones = 0
    streamtraces = []
    
    with open('streamtraces-3d-3.dat', 'r') as f:
        
        
        for line in f:
            
            if 'ZONE' in line:
                
                nZones += 1
                
                f.readline()
                npntsline = f.readline()
                nPoints = int(re.search('I=(\d+)', npntsline)[1])
                
                f.readline()
                f.readline()
                coords = np.zeros((nPoints, 3))
                for i in range(nPoints):
                    coords[i, :] = [float(x) for x in f.readline().split()]
                
                streamtraces.append(coords)
    
    
    return streamtraces
    
                
stream = read_streamtraces_file()
```

```python
import matplotlib.pyplot as plt
fig = plt.figure()
ax = plt.subplot(111)
for i, s in enumerate(stream):
    ax.plot(s[:, 0], s[:, 1], color=[0.3, 0.3, 0.3], alpha=0.5)
```
