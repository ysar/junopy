```python
import numpy as np
import pandas as pd
from scipy.constants import epsilon_0, m_e, e, m_p, c, mu_0
```

```python
columns = ['duration', 'distance', 'localtime', 'cutoff', 'density', 'mag', 'ioninertial', 'alfvenspeed', 'diameter']
df = pd.DataFrame(index=[0, 1], columns=columns)
```

```python
df['duration'][0] = 22
df['duration'][1] = 62

df['distance'][0] = 74
df['distance'][1] = 92

df['localtime'][0] = 4.05
df['localtime'][1] = 3.55

df['mag'][0] = 5
df['mag'][1] = 4.5

df['cutoff'][0] = 500 * 2 * np.pi
df['cutoff'][1] = 1000 * 2 * np.pi

df['density'] = df['cutoff']**2 * epsilon_0 * m_e / e**2 / 1e6
df['ioninertial'] = c / (df['cutoff'] * np.sqrt(m_e / m_p / 16)) / 1e3
df['alfvenspeed'] = df['mag'] / np.sqrt(mu_0 * 16 * m_p * 1e6) / df['density'].pow(0.5) / 1000 * 1e-9

df['diameter'] = df['alfvenspeed'] * df['duration'] 
df['diameter-r'] = df['diameter'] / 71492
df['ioninertial-r'] = df['ioninertial'] / 71492
df['dia/ion-inertial'] = df['diameter'] / df['ioninertial']
```
