```python
je1 = jedi.JediData(filename='../data/jedi/JED_180_HIERSESP_CDR_2017076_V04.TAB', 
                    species='ELECTRON',
                    start=dt.datetime.strptime('2017-076T05:30:00', '%Y-%jT%H:%M:%S'),
                    end=dt.datetime.strptime('2017-076T07:00:00', '%Y-%jT%H:%M:%S'),
                    dtbin='30S')

je2 = jedi.JediData(filename='../data/jedi/JED_270_HIERSESP_CDR_2017076_V04.TAB', 
                    species='ELECTRON',
                    start=dt.datetime.strptime('2017-076T05:30:00', '%Y-%jT%H:%M:%S'),
                    end=dt.datetime.strptime('2017-076T07:00:00', '%Y-%jT%H:%M:%S'),
                    dtbin='30S')

jproton = jedi.JediData(filename='../data/jedi/JED_090_LOERSTOFXER_CDR_2017076_V04.TAB', 
                    species='PROTON',
                    start=dt.datetime.strptime('2017-076T05:30:00', '%Y-%jT%H:%M:%S'),
                    end=dt.datetime.strptime('2017-076T07:00:00', '%Y-%jT%H:%M:%S'),
                    dtbin='30S')

je3 = jedi.merge_two_jedi_data(je1, je2)

fig, axs = plt.subplots(4, 1, dpi=300, figsize=(9,6), sharex=True)


je3.plot_energy_spectra(axs[0], cmap='jet', vmin=1, vmax=3e3)
je3.plot_pitchangle_spectra(axs[1], cmap='jet', vmin=1, vmax=3e3)
jproton.plot_energy_spectra(axs[2], cmap='jet', vmin=5e-3, vmax=2e3)
jproton.plot_pitchangle_spectra(axs[3], cmap='jet', vmin=0.4, vmax=2e3)

fig.suptitle('2017-076, Electron and Proton Flux, Units - counts/s/cm$^2$/str/keV', y=0.91)
for a in axs:
    a.label_outer()

fig.savefig('../out_images/2017-076_JEDISpectra_PDS.png', bbox_inches='tight', facecolor='white')
plt.show()
```
