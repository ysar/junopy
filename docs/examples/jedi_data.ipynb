```python
import datetime as dt
import numpy as np
from matplotlib import pyplot as plt

from junopy import config
from junopy import index
from junopy import jedi
from junopy import utils

config.ConfigVars.set_root("../.junodata")
```

```python
start = dt.datetime(2020, 1, 1, 2)
stop = dt.datetime(2020, 1, 1, 23)
jedi_index = index.IndexFile(dataset='jedi')
jedi_files = jedi_index.get_files_between_dates(start, stop, sid="ERSESP", do_download=True, get_label=False)
jedi_files
```

```python
jedi_data = jedi.JediData(filename=jedi_files[0].with_suffix(".hdf5"), start=start, stop=stop, species="ELECTRON")
```
