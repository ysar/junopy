```python
import numpy as np
import pandas as pd
from scipy.constants import epsilon_0, m_e, e, m_p, c, mu_0


def calc_elec_density(cutoff):
    return (cutoff * 2 * np.pi)**2 * epsilon_0 * m_e / e**2 / 1e6
    
    
def calc_ion_inertial(cutoff, ionmass=16):
    return c / ((cutoff * 2 * np.pi) * np.sqrt(m_e / m_p / ionmass)) / 1e3
```

```python
calc_elec_density(700)
```
