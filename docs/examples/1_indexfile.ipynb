# Load an index file

This notebook uses the `junopy.index` module to load the index files for the magnetometer and waves survey
data that is stored on the PDS.

```python
import numpy as np
import datetime as dt

from junopy import index
from junopy import config

# Set the junopy root directory to wherever you want the data to be downloaded.
config.ConfigVars.set_root(".junodata")   # <---- Modify as necessary.
```

Let's try to find relevant magnetometer and Waves data between two times

```python
start = dt.datetime(2020, 1, 1, 2)
stop = dt.datetime(2020, 1, 1, 23)
```

## Load the magnetometer index file

The following code will load the magnetometer index file, downloading it if it is not already present.  Each `IndexFile` object 
contains a `df` attribute that is a `pandas` `DataFrame` and contains relevant information stored in the INDEX.TAB file for that 
PDS archive.

```python
mag_index = index.IndexFile(dataset='mag')
mag_index.df.head()
```

We can use the index files to get a list of relevant data files that correspond to the particular SID using the 
`get_files_between_dates` method.

```python
mag_files = mag_index.get_files_between_dates(start, stop, sid="SS 1 SECOND")
mag_files
```

> ℹ️ By default, the `get_files_between_dates` only returns a list of paths to where the data file *could* be 
found, if it was downloaded.  To have this method actually download the data, pass in `do_download=True` 
argument.

> ℹ️ The magnetometer data file here appears to have the suffix of `.sts`, which is an ASCII text file.
> Once that file is loaded using a `MagData` instance, `junopy` will convert any magnetometer ASCII `.sts`
> file into an HDF5 file with format `.hdf5`.  This saves space on disk and shortens loading times with no loss of data.

## Load the Waves index file for survey data

Likewise, we can repeat this process for the Waves survey dataset.

```python
waves_index = index.IndexFile(dataset='waves-survey')
waves_index.df.head()
```

```python
waves_files = waves_index.get_files_between_dates(start, stop, sid="SURVEY")
waves_files
```

Where you can see both electric and magnetic field spectra were recognized.  Like the magnetometer data,
Waves survey data files, which are `.CSV` files by default, as also converted into `.hdf5` files once they have 
been read using a `WavesSurveyData`.
